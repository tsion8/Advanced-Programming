```python
"""
SORTING ALGORITHMS COMPARISON IN BIOINFORMATICS
================================================

Merge Sort | Quick Sort | Heap Sort | Radix Sort

Comparison based on:
- Working principle
- Time complexity
- Space complexity
- Stability
- Advantages
- Disadvantages
- Bioinformatics applications
"""

# ==========================================================
# COMPARISON TABLE
# ==========================================================

def print_comparison_table():

    headers = [
        "Criterion",
        "Merge Sort",
        "Quick Sort",
        "Heap Sort",
        "Radix Sort"
    ]

    rows = [

        ["Working Principle",
         "Divide-and-conquer:\nsplit array and merge\nsorted sub-arrays",
         "Divide-and-conquer:\nchoose pivot and\npartition elements",
         "Build binary heap,\nextract maximum element\nrepeatedly",
         "Non-comparison sort:\nsort digits/characters\nusing counting sort"],


        ["Best Time Complexity",
         "O(n log n)",
         "O(n log n)",
         "O(n log n)",
         "O(nk)"],


        ["Average Time Complexity",
         "O(n log n)",
         "O(n log n)",
         "O(n log n)",
         "O(nk)"],


        ["Worst Time Complexity",
         "O(n log n)",
         "O(n²)",
         "O(n log n)",
         "O(nk)"],


        ["Space Complexity",
         "O(n)",
         "O(log n)*",
         "O(1)*",
         "O(n+k)"],


        ["Stability",
         "Stable",
         "Not stable",
         "Not stable",
         "Stable"],


        ["Advantages",
         "Guaranteed performance;\nstable;\nexcellent for large data",
         "Very fast in practice;\ncache-friendly;\nlow overhead",
         "Guaranteed O(n log n);\nin-place sorting;\nlow memory usage",
         "Very fast when key length\n(k) is small;\npredictable performance"],


        ["Disadvantages",
         "Requires extra memory;\nslower than quicksort\nfor small datasets",
         "Worst-case O(n²);\npivot selection affects\nperformance",
         "Poor cache locality;\nnot stable",
         "Only works with\nstructured data;\nrequires extra memory"],


        ["Bioinformatics Applications",
         "Large genomic datasets;\nFASTQ/BAM external sorting;\nchromosome coordinate sorting",
         "Gene expression values;\nalignment scores;\np-values in genomics",
         "Priority queues in biological\nnetwork analysis;\ntop-k genes and variants",
         "DNA k-mers;\nsequence indexing;\nfixed-length barcodes"]
    ]


    label_width = 25
    col_width = 35


    def print_row(cells):

        split_cells = [
            str(cell).split("\n")
            for cell in cells
        ]

        max_lines = max(
            len(cell)
            for cell in split_cells
        )


        for i in range(max_lines):

            row = []

            for j, cell in enumerate(split_cells):

                text = cell[i] if i < len(cell) else ""

                width = label_width if j == 0 else col_width

                row.append(text.ljust(width))

            print(" | ".join(row))


    print("\n")
    print("=" * 170)
    print("SORTING ALGORITHMS COMPARISON")
    print("=" * 170)


    print_row(headers)

    print("-" * 170)


    for row in rows:

        print_row(row)

        print("-" * 170)



    print("""
NOTE:
* O(log n) for Quick Sort and O(1) for Heap Sort represent
  the theoretical algorithm complexity.

* The Python implementations below use additional memory because
  temporary lists and copies are created for easier implementation.
""")


print_comparison_table()



# ==========================================================
# SORTING ALGORITHM IMPLEMENTATIONS
# ==========================================================


# -----------------------------
# Merge Sort
# -----------------------------

def merge(left, right):

    result = []

    i = 0
    j = 0


    while i < len(left) and j < len(right):

        if left[i] <= right[j]:

            result.append(left[i])

            i += 1

        else:

            result.append(right[j])

            j += 1


    result.extend(left[i:])

    result.extend(right[j:])


    return result



def merge_sort(arr):

    if len(arr) <= 1:

        return arr


    mid = len(arr) // 2


    left = merge_sort(arr[:mid])

    right = merge_sort(arr[mid:])


    return merge(left, right)



# -----------------------------
# Quick Sort
# -----------------------------

def quick_sort(arr):

    if len(arr) <= 1:

        return arr


    pivot = arr[len(arr)//2]


    left = [
        x for x in arr
        if x < pivot
    ]


    middle = [
        x for x in arr
        if x == pivot
    ]


    right = [
        x for x in arr
        if x > pivot
    ]


    return (
        quick_sort(left)
        + middle
        + quick_sort(right)
    )



# -----------------------------
# Heap Sort
# -----------------------------

def heap_sort(arr):

    import heapq


    heap = arr.copy()


    heapq.heapify(heap)


    sorted_list = []


    while heap:

        sorted_list.append(
            heapq.heappop(heap)
        )


    return sorted_list



# -----------------------------
# Radix Sort
# -----------------------------

def counting_sort_for_radix(arr, exp):

    n = len(arr)

    output = [0] * n

    count = [0] * 10


    for number in arr:

        index = (
            number // exp
        ) % 10

        count[index] += 1



    for i in range(1,10):

        count[i] += count[i-1]



    for i in range(n-1,-1,-1):

        index = (
            arr[i] // exp
        ) % 10


        output[
            count[index]-1
        ] = arr[i]


        count[index] -= 1



    return output



def radix_sort(arr):

    if not arr:

        return arr


    arr = arr.copy()


    maximum = max(arr)


    exp = 1


    while maximum // exp > 0:

        arr = counting_sort_for_radix(
            arr,
            exp
        )

        exp *= 10



    return arr



# ==========================================================
# DEMONSTRATION
# ==========================================================


sample_data = [
    38, 27, 43,
    3, 9, 82, 10
]


print("\nDEMONSTRATION")
print("=" * 50)

print("Original Data:")
print(sample_data)


print("\nSorted Results:")

print(
    "Merge Sort :",
    merge_sort(sample_data)
)


print(
    "Quick Sort :",
    quick_sort(sample_data)
)


print(
    "Heap Sort  :",
    heap_sort(sample_data)
)


print(
    "Radix Sort :",
    radix_sort(sample_data)
)
```

```python
"""
QUESTION 2: WHY IS BINARY SEARCH SIGNIFICANTLY FASTER
THAN LINEAR SEARCH?
======================================================
"""

print("""

QUESTION 2: WHY IS BINARY SEARCH SIGNIFICANTLY FASTER
THAN LINEAR SEARCH?
======================================================


1. WHY BINARY SEARCH IS FASTER
===============================

Linear Search:
--------------
Linear search checks every element one by one until the target is found.
In the worst case, it may need to check all n elements.

Time Complexity:
    Best Case:    O(1)
    Average Case: O(n)
    Worst Case:   O(n)


Binary Search:
--------------
Binary search repeatedly divides the search space into two halves.

Example:

Initial data:
    n

After first comparison:
    n/2

After second comparison:
    n/4

After third comparison:
    n/8

The search continues until only one element remains.

Time Complexity:
    Best Case:    O(1)
    Average Case: O(log n)
    Worst Case:   O(log n)


Why it is faster:
-----------------

For a dataset of 1,000,000 elements:

Linear Search:
    Up to 1,000,000 comparisons

Binary Search:
    Approximately 20 comparisons

Therefore, Binary Search becomes much faster as dataset size increases.



2. HOW BINARY SEARCH WORKS
============================

Binary Search follows these steps:

Step 1:
Set two boundaries:

    low  = first element index
    high = last element index


Step 2:
Find the middle element:

    mid = (low + high) // 2


Step 3:
Compare the middle value with the target.

If:

    target == middle element:
        Element is found

    target > middle element:
        Search the right half

    target < middle element:
        Search the left half


Step 4:
Repeat until the element is found or the search range becomes empty.



3. WHY BINARY SEARCH REQUIRES SORTED DATA
===========================================

Binary Search depends on the order of the data.

Example of sorted data:

[10, 20, 30, 40, 50, 60]


If the middle value is 30 and the target is 50:

We know the target must be on the right side.

The left half can be discarded safely.


However, with unsorted data:

[40, 10, 60, 20, 50, 30]


The middle value gives no information about where the target exists.

Therefore, half of the data cannot be removed.

Binary Search only works efficiently when the data is sorted.



4. SITUATIONS WHERE BINARY SEARCH CANNOT BE USED EFFECTIVELY
==============================================================


1. Unsorted Data
----------------
Binary Search requires sorted input.
If the data is not sorted, it must be sorted first.


2. Frequently Changing Data
---------------------------
If data is constantly inserted or deleted, maintaining sorted order
may become expensive.


3. Linked Lists
---------------
Binary Search requires direct access to the middle element.

Arrays allow:

    Access middle element immediately

Linked lists require:

    Traversing nodes one by one

Therefore, Binary Search loses its advantage.


4. Very Small Datasets
----------------------
For small datasets, Linear Search may be faster because it has
less overhead.


5. Similarity or Approximate Searching
--------------------------------------
Binary Search is designed for exact matches.

It cannot directly perform tasks such as:

- Finding similar DNA sequences
- Finding related genes
- Searching approximate matches



5. BIOINFORMATICS APPLICATIONS
===============================

Binary Search is useful in:

- Searching sorted genomic coordinates in BAM files
- Finding variants in sorted VCF files
- Searching indexed DNA sequence databases
- Locating k-mers in sorted sequence indexes

""")



# ======================================================
# CODE DEMONSTRATION
# ======================================================


def linear_search(arr, target):

    comparisons = 0

    for i, value in enumerate(arr):

        comparisons += 1

        if value == target:

            return i, comparisons

    return -1, comparisons



def binary_search(arr, target):

    low = 0
    high = len(arr)-1

    comparisons = 0


    while low <= high:

        mid = (low + high)//2

        comparisons += 1


        if arr[mid] == target:

            return mid, comparisons


        elif arr[mid] < target:

            low = mid + 1


        else:

            high = mid - 1


    return -1, comparisons



# Demonstration

data = list(range(0,1000,2))

target = 842


linear_result = linear_search(data, target)

binary_result = binary_search(data, target)



print("\nDEMONSTRATION")
print("============================")

print("Dataset size:", len(data))

print("\nLinear Search")
print("Index:", linear_result[0])
print("Comparisons:", linear_result[1])


print("\nBinary Search")
print("Index:", binary_result[0])
print("Comparisons:", binary_result[1])
```

```python
"""
QUESTION 5: MANUAL MERGE SORT ON DNA SEQUENCE LENGTHS
=========================================================

Input list:
    [560, 210, 340, 890, 450, 120, 780]


STEP 1: DIVISION (splitting down to single elements)
==========================================================

    Original: [560, 210, 340, 890, 450, 120, 780]   (7 elements)

    Split into:
        Left  = [560, 210, 340]
        Right = [890, 450, 120, 780]

    --- Splitting Left = [560, 210, 340] ---
        Split into:
            Left  = [560]
            Right = [210, 340]

        [560] -> single element, cannot split further (base case)

        --- Splitting [210, 340] ---
            Split into:
                Left  = [210]
                Right = [340]
            [210] -> base case
            [340] -> base case

    --- Splitting Right = [890, 450, 120, 780] ---
        Split into:
            Left  = [890, 450]
            Right = [120, 780]

        --- Splitting [890, 450] ---
            Split into:
                Left  = [890]
                Right = [450]
            [890] -> base case
            [450] -> base case

        --- Splitting [120, 780] ---
            Split into:
                Left  = [120]
                Right = [780]
            [120] -> base case
            [780] -> base case


    Full division tree:

    [560,210,340,890,450,120,780]
    ├── [560,210,340]
    │     ├── [560]
    │     └── [210,340]
    │           ├── [210]
    │           └── [340]
    └── [890,450,120,780]
          ├── [890,450]
          │     ├── [890]
          │     └── [450]
          └── [120,780]
                ├── [120]
                └── [780]


STEP 2: MERGE STEPS (combining back in sorted order)
==========================================================

    Merge 1: [210] + [340]
        Compare 210 vs 340 -> 210 is smaller -> take 210
        Only 340 left -> take 340
        Result: [210, 340]

    Merge 2: [560] + [210, 340]
        Compare 560 vs 210 -> 210 is smaller -> take 210
        Compare 560 vs 340 -> 340 is smaller -> take 340
        Only 560 left -> take 560
        Result: [210, 340, 560]

    Merge 3: [890] + [450]
        Compare 890 vs 450 -> 450 is smaller -> take 450
        Only 890 left -> take 890
        Result: [450, 890]

    Merge 4: [120] + [780]
        Compare 120 vs 780 -> 120 is smaller -> take 120
        Only 780 left -> take 780
        Result: [120, 780]

    Merge 5: [450, 890] + [120, 780]
        Compare 450 vs 120 -> 120 is smaller -> take 120
        Compare 450 vs 780 -> 450 is smaller -> take 450
        Compare 890 vs 780 -> 780 is smaller -> take 780
        Only 890 left -> take 890
        Result: [120, 450, 780, 890]

    Merge 6 (FINAL MERGE): [210, 340, 560] + [120, 450, 780, 890]
        Compare 210 vs 120 -> 120 is smaller -> take 120
        Compare 210 vs 450 -> 210 is smaller -> take 210
        Compare 340 vs 450 -> 340 is smaller -> take 340
        Compare 560 vs 450 -> 450 is smaller -> take 450
        Compare 560 vs 780 -> 560 is smaller -> take 560
        Only [780, 890] left -> take 780, then 890
        Result: [120, 210, 340, 450, 560, 780, 890]


FINAL SORTED LIST
==========================================================
    [120, 210, 340, 450, 560, 780, 890]
"""

print(__doc__)


# ---- Code to verify the manual trace ----

def merge_sort_verbose(arr, depth=0):
    indent = "  " * depth
    if len(arr) <= 1:
        print(f"{indent}Base case: {arr}")
        return arr

    mid = len(arr) // 2
    left_part = arr[:mid]
    right_part = arr[mid:]
    print(f"{indent}Split {arr} -> {left_part} | {right_part}")

    left = merge_sort_verbose(left_part, depth + 1)
    right = merge_sort_verbose(right_part, depth + 1)

    merged = merge_verbose(left, right, depth)
    return merged


def merge_verbose(left, right, depth):
    indent = "  " * depth
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    print(f"{indent}Merge {left} + {right} -> {result}")
    return result


sequence_lengths = [560, 210, 340, 890, 450, 120, 780]

print("Trace of Merge Sort:")
print("=" * 50)
final_result = merge_sort_verbose(sequence_lengths)

print("\nFinal sorted list:", final_result)
```

```python
"""
QUESTION 6: MANUAL QUICK SORT ON DNA SEQUENCE LENGTHS
=========================================================

Input list:
    [560, 210, 340, 890, 450, 120, 780]

Pivot strategy used: LAST ELEMENT of each sub-array
Partition scheme: Lomuto partition scheme


CALL 1: quicksort([560, 210, 340, 890, 450, 120, 780])
==========================================================

Pivot selection:
    Pivot = last element = 780

Partitioning:
    Array: [560, 210, 340, 890, 450, 120, 780]
    i = -1 (tracks boundary of elements smaller than pivot)

    j=0: 560 < 780? Yes -> i=0, swap arr[0] with arr[0] (no change)
         Array: [560, 210, 340, 890, 450, 120, 780]

    j=1: 210 < 780? Yes -> i=1, swap arr[1] with arr[1] (no change)
         Array: [560, 210, 340, 890, 450, 120, 780]

    j=2: 340 < 780? Yes -> i=2, swap arr[2] with arr[2] (no change)
         Array: [560, 210, 340, 890, 450, 120, 780]

    j=3: 890 < 780? No -> skip

    j=4: 450 < 780? Yes -> i=3, swap arr[3] with arr[4]
         Array: [560, 210, 340, 450, 890, 120, 780]

    j=5: 120 < 780? Yes -> i=4, swap arr[4] with arr[5]
         Array: [560, 210, 340, 450, 120, 890, 780]

    Final step: swap arr[i+1] (index 5) with pivot (index 6)
         Array: [560, 210, 340, 450, 120, 780, 890]

    Pivot 780 now sits at its correct final position: index 5

    Result of partition:
        Left  = [560, 210, 340, 450, 120]   (all < 780)
        Pivot = 780                          (fixed in place)
        Right = [890]                        (all > 780)


CALL 2: quicksort([560, 210, 340, 450, 120])
==========================================================

Pivot selection:
    Pivot = last element = 120

Partitioning:
    Array: [560, 210, 340, 450, 120]
    i = -1

    j=0: 560 < 120? No -> skip
    j=1: 210 < 120? No -> skip
    j=2: 340 < 120? No -> skip
    j=3: 450 < 120? No -> skip

    No elements smaller than pivot found, so i stays at -1.

    Final step: swap arr[i+1] (index 0) with pivot (index 4)
         Array: [120, 210, 340, 450, 560]

    Pivot 120 now sits at index 0.

    Result of partition:
        Left  = []                          (empty, nothing < 120)
        Pivot = 120
        Right = [210, 340, 450, 560]


CALL 3: quicksort([210, 340, 450, 560])
==========================================================

Pivot selection:
    Pivot = last element = 560

Partitioning:
    Array: [210, 340, 450, 560]
    i = -1

    j=0: 210 < 560? Yes -> i=0, swap arr[0] with arr[0] (no change)
    j=1: 340 < 560? Yes -> i=1, swap arr[1] with arr[1] (no change)
    j=2: 450 < 560? Yes -> i=2, swap arr[2] with arr[2] (no change)

    Final step: swap arr[i+1] (index 3) with pivot (index 3) -- no change
         Array: [210, 340, 450, 560]

    Pivot 560 sits at index 3.

    Result of partition:
        Left  = [210, 340, 450]
        Pivot = 560
        Right = []                          (empty)


CALL 4: quicksort([210, 340, 450])
==========================================================

Pivot selection:
    Pivot = last element = 450

Partitioning:
    Array: [210, 340, 450]
    i = -1

    j=0: 210 < 450? Yes -> i=0, swap arr[0] with arr[0] (no change)
    j=1: 340 < 450? Yes -> i=1, swap arr[1] with arr[1] (no change)

    Final step: swap arr[i+1] (index 2) with pivot (index 2) -- no change
         Array: [210, 340, 450]

    Pivot 450 sits at index 2.

    Result of partition:
        Left  = [210, 340]
        Pivot = 450
        Right = []


CALL 5: quicksort([210, 340])
==========================================================

Pivot selection:
    Pivot = last element = 340

Partitioning:
    Array: [210, 340]
    i = -1

    j=0: 210 < 340? Yes -> i=0, swap arr[0] with arr[0] (no change)

    Final step: swap arr[i+1] (index 1) with pivot (index 1) -- no change
         Array: [210, 340]

    Pivot 340 sits at index 1.

    Result of partition:
        Left  = [210]
        Pivot = 340
        Right = []


CALL 6: quicksort([210])
==========================================================
    Base case (single element) -> already sorted: [210]


REASSEMBLING THE RESULTS (bottom-up)
==========================================================
    quicksort([210])                       -> [210]
    quicksort([210,340])                   -> [210] + [340]           = [210, 340]
    quicksort([210,340,450])               -> [210,340] + [450]       = [210, 340, 450]
    quicksort([210,340,450,560])           -> [210,340,450] + [560]   = [210, 340, 450, 560]
    quicksort([560,210,340,450,120])       -> [120] + [210,340,450,560] = [120, 210, 340, 450, 560]
    quicksort([560,210,340,890,450,120,780]) -> [120,210,340,450,560] + [780] + [890]


FINAL SORTED LIST
==========================================================
    [120, 210, 340, 450, 560, 780, 890]
"""

print(__doc__)


# ---- Code to verify the manual trace ----

def quick_sort_verbose(arr, depth=0):
    indent = "  " * depth
    if len(arr) <= 1:
        print(f"{indent}Base case: {arr}")
        return arr

    pivot = arr[-1]
    print(f"{indent}Pivot selected: {pivot} (from {arr})")

    i = -1
    arr = arr.copy()
    for j in range(len(arr) - 1):
        if arr[j] < pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[-1] = arr[-1], arr[i + 1]
    pivot_index = i + 1
    print(f"{indent}After partitioning around pivot {pivot}: {arr} "
          f"(pivot fixed at index {pivot_index})")

    left_part = arr[:pivot_index]
    right_part = arr[pivot_index + 1:]
    print(f"{indent}Left = {left_part}, Pivot = {pivot}, Right = {right_part}")

    sorted_left = quick_sort_verbose(left_part, depth + 1)
    sorted_right = quick_sort_verbose(right_part, depth + 1)

    result = sorted_left + [pivot] + sorted_right
    print(f"{indent}Combined result: {result}")
    return result


sequence_lengths = [560, 210, 340, 890, 450, 120, 780]

print("Trace of Quick Sort:")
print("=" * 50)
final_result = quick_sort_verbose(sequence_lengths)

print("\nFinal sorted list:", final_result)
```

```python
"""
QUESTION 7: BINARY SEARCH ON SORTED GENE POSITIONS
==================================================

Given Sorted Gene Positions:

[105, 212, 315, 427, 532, 610, 745, 899, 1040]

Target Value:
610


STEP 1: INITIAL SEARCH
==================================================

Low  = 0
High = 8

Middle Index:

    mid = (0 + 8) // 2 = 4

Middle Element:

    532

Comparison:

    610 > 532

Decision:

    Search the RIGHT half.

New Search Range:

    Low  = 5
    High = 8


STEP 2: SECOND ITERATION
==================================================

Current Search Range:

    [610, 745, 899, 1040]

Low  = 5
High = 8

Middle Index:

    mid = (5 + 8) // 2 = 6

Middle Element:

    745

Comparison:

    610 < 745

Decision:

    Search the LEFT half.

New Search Range:

    Low  = 5
    High = 5


STEP 3: THIRD ITERATION
==================================================

Current Search Range:

    [610]

Low  = 5
High = 5

Middle Index:

    mid = (5 + 5) // 2 = 5

Middle Element:

    610

Comparison:

    610 == 610

Result:

    Target Found!


FINAL RESULT
==================================================

Target Gene Position : 610

Found at Index      : 5

Number of Iterations: 3
"""

print(__doc__)


# --------------------------------------------------
# Code Verification
# --------------------------------------------------

def binary_search_verbose(arr, target):

    low = 0
    high = len(arr) - 1
    iteration = 1

    while low <= high:

        print(f"\nIteration {iteration}")
        print("-" * 40)

        mid = (low + high) // 2

        print(f"Low   = {low}")
        print(f"High  = {high}")
        print(f"Middle Index = {mid}")
        print(f"Middle Value = {arr[mid]}")

        if arr[mid] == target:
            print("\nTarget Found!")
            print(f"Index = {mid}")
            print(f"Iterations = {iteration}")
            return mid

        elif arr[mid] < target:
            print(f"{target} > {arr[mid]}")
            print("Move to RIGHT half")
            low = mid + 1

        else:
            print(f"{target} < {arr[mid]}")
            print("Move to LEFT half")
            high = mid - 1

        iteration += 1

    print("\nTarget Not Found")
    return -1


gene_positions = [105, 212, 315, 427, 532, 610, 745, 899, 1040]

target = 610

print("\nBinary Search Trace")
print("=" * 50)

binary_search_verbose(gene_positions, target)
```

```python
"""
QUESTION 8: LOWER BOUND SEARCH AND UPPER BOUND SEARCH
=====================================================

Given Sorted List:

[10, 20, 20, 20, 35, 40, 50]

Target Value:
20


LOWER BOUND SEARCH (First Occurrence)
=====================================================

Definition:
The Lower Bound Search finds the first occurrence of the target
value in a sorted list.

Search Process:

Iteration 1

Low  = 0
High = 6

Middle Index = (0 + 6) // 2 = 3

Middle Value = 20

Target found.

Continue searching LEFT half to find the first occurrence.

High = 2


Iteration 2

Low  = 0
High = 2

Middle Index = (0 + 2) // 2 = 1

Middle Value = 20

Target found.

Continue searching LEFT half.

High = 0


Iteration 3

Low  = 0
High = 0

Middle Index = (0 + 0) // 2 = 0

Middle Value = 10

10 < 20

Search RIGHT half.

Low = 1


Search Ends

First occurrence of 20 is at Index 1.



UPPER BOUND SEARCH (Last Occurrence)
=====================================================

Definition:
The Upper Bound Search finds the last occurrence of the target
value in a sorted list.

Search Process:

Iteration 1

Low  = 0
High = 6

Middle Index = 3

Middle Value = 20

Target found.

Continue searching RIGHT half.

Low = 4


Iteration 2

Low  = 4
High = 6

Middle Index = 5

Middle Value = 40

40 > 20

Search LEFT half.

High = 4


Iteration 3

Low  = 4
High = 4

Middle Index = 4

Middle Value = 35

35 > 20

Search LEFT half.

High = 3


Search Ends

Last occurrence of 20 is at Index 3.



FINAL RESULTS
=====================================================

First occurrence of 20 : Index 1

Last occurrence of 20  : Index 3
"""

print(__doc__)


# =====================================================
# Code Verification
# =====================================================

def lower_bound(arr, target):

    low = 0
    high = len(arr) - 1

    result = -1
    iteration = 1

    print("\nLOWER BOUND SEARCH")
    print("=" * 50)

    while low <= high:

        mid = (low + high) // 2

        print(f"\nIteration {iteration}")
        print(f"Low = {low}, High = {high}, Mid = {mid}")
        print(f"Middle Value = {arr[mid]}")

        if arr[mid] == target:
            result = mid
            print("Target found -> Continue LEFT")
            high = mid - 1

        elif arr[mid] < target:
            print("Search RIGHT")
            low = mid + 1

        else:
            print("Search LEFT")
            high = mid - 1

        iteration += 1

    return result



def upper_bound(arr, target):

    low = 0
    high = len(arr) - 1

    result = -1
    iteration = 1

    print("\nUPPER BOUND SEARCH")
    print("=" * 50)

    while low <= high:

        mid = (low + high) // 2

        print(f"\nIteration {iteration}")
        print(f"Low = {low}, High = {high}, Mid = {mid}")
        print(f"Middle Value = {arr[mid]}")

        if arr[mid] == target:
            result = mid
            print("Target found -> Continue RIGHT")
            low = mid + 1

        elif arr[mid] < target:
            print("Search RIGHT")
            low = mid + 1

        else:
            print("Search LEFT")
            high = mid - 1

        iteration += 1

    return result


# -----------------------------------------------------

numbers = [10, 20, 20, 20, 35, 40, 50]

target = 20

first = lower_bound(numbers, target)
last = upper_bound(numbers, target)

print("\nFINAL RESULTS")
print("=" * 50)
print(f"First occurrence of {target}: Index {first}")
print(f"Last occurrence of {target}: Index {last}")
```

```python
"""
QUESTION 9: ANALYZING WHOLE GENOME SEQUENCING (WGS) DATA
========================================================

Scenario:
You are developing software to analyze Whole Genome Sequencing
(WGS) data containing:

• 30 million variants
• 5,000 samples

Answer the following questions.


a) Which sorting algorithm would you choose?
========================================================

Recommended Algorithm:
    Merge Sort

Justification:
    • Merge Sort has a guaranteed time complexity of O(n log n).
    • It is stable, meaning equal elements retain their original order.
    • It performs well on very large datasets.
    • It is suitable for external sorting when the data cannot fit
      completely into memory.
    • Variant files (VCF/BAM) are often extremely large, making
      Merge Sort a reliable choice.

Although Quick Sort is usually faster in memory, Merge Sort is
preferred for very large genomic datasets because of its
predictable performance and stability.



b) Which searching algorithm would you choose?
========================================================

Recommended Algorithm:
    Binary Search

Justification:
    • Variant positions are usually stored in sorted order.
    • Binary Search has O(log n) time complexity.
    • It repeatedly divides the search space in half.
    • Even with millions of variants, only a small number of
      comparisons are required.

Example:

30,000,000 variants

Linear Search:
    Up to 30,000,000 comparisons

Binary Search:
    Approximately log₂(30,000,000) ≈ 25 comparisons



c) How would you optimize the software?
========================================================

Several optimization techniques can improve performance.

1. Parallel Processing
   Process multiple samples simultaneously using multiple CPU cores.

2. Multi-threading
   Analyze different chromosomes or genomic regions in parallel.

3. Efficient Algorithms
   Use optimized alignment algorithms such as BWA or Minimap2.

4. Compressed Data
   Store genomic files in compressed formats such as BAM or CRAM.

5. Memory Optimization
   Load only the required genomic regions instead of the entire genome.

6. Caching
   Store frequently accessed genomic regions in memory.



d) Which data structures would improve performance?
========================================================

Several data structures are useful for WGS analysis.

Hash Tables
    • Fast lookup of variants by chromosome and position.
    • Average lookup time: O(1).

Binary Search Trees (BST)
    • Efficient insertion and searching.
    • Useful for maintaining sorted data.

Heaps
    • Efficient priority queues.
    • Useful when selecting top-quality variants.

Arrays
    • Efficient storage of genomic positions.
    • Fast random access.

Interval Trees
    • Efficient searching of overlapping genomic regions.
    • Widely used in genome annotation.

Hash Maps
    • Fast mapping of gene names to genomic information.



e) Explain how indexing could reduce computation time
========================================================

Indexing creates a searchable data structure before analysis begins.

Instead of scanning every variant one by one, the software directly
jumps to the required genomic location.

Examples:

• BWA and Bowtie use the Burrows-Wheeler Transform (BWT)
  together with the FM-index.

• BAM files use index files (.bai).

• VCF files use Tabix indexes (.tbi).

Without an index:

    Search Time ≈ O(n)

With an index:

    Search Time ≈ O(log n) or better,
    depending on the indexing structure.

For datasets containing 30 million variants, indexing dramatically
reduces computation time and allows rapid retrieval of specific
genomic regions.



SUMMARY
========================================================

Sorting Algorithm
    Merge Sort

Searching Algorithm
    Binary Search

Optimization
    Parallel processing, multithreading, efficient algorithms,
    compression, caching, memory optimization

Useful Data Structures
    Hash Tables
    Arrays
    Binary Search Trees
    Heaps
    Interval Trees
    Hash Maps

Indexing
    Uses structures such as FM-index, BWT, BAM index (.bai),
    and Tabix (.tbi) to avoid scanning the entire dataset,
    greatly reducing search time.
"""

print(__doc__)


# ==========================================================
# Demonstration Code
# ==========================================================

import math

variants = 30_000_000

linear_search = variants
binary_search = math.ceil(math.log2(variants))

print("WGS DATA ANALYSIS")
print("=" * 50)
print(f"Number of variants : {variants:,}")
print(f"Number of samples  : 5,000\n")

print("Estimated Search Comparisons")
print("-" * 50)
print(f"Linear Search : {linear_search:,} comparisons")
print(f"Binary Search : {binary_search} comparisons\n")

print("Recommended Choices")
print("-" * 50)
print("Sorting Algorithm  : Merge Sort")
print("Searching Algorithm: Binary Search")
print("Indexing           : FM-index / BWT / BAM(.bai) / Tabix(.tbi)")
```
