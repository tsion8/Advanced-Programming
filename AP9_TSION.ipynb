```python
# QUESTION 1 — DEFINITIONS

print("=" * 60)
print("QUESTION 1 — DEFINITIONS")
print("=" * 60)

print("\nA. String")
print("A string is a finite sequence of characters.")
print('Example: "ACGT"')

print("\nB. Prefix")
print("A prefix is a substring that starts at the beginning of a string.")
print('Example: Prefixes of "ACGT" are A, AC, ACG, ACGT')

print("\nC. Suffix")
print("A suffix is a substring that ends at the end of a string.")
print('Example: Suffixes of "ACGT" are T, GT, CGT, ACGT')

print("\nD. Substring")
print("A substring is a continuous sequence of characters within a string.")
print('Example: "CG" is a substring of "ACGT"')

print("\nE. Pattern")
print("A pattern is a string that is searched for inside a larger text.")
print('Example: "CGT" is a pattern searched in "ACGTACGT"')

print("\nF. Alphabet")
print("An alphabet is the set of symbols used to form strings.")
print("DNA alphabet = {A, C, G, T}")

print("\nG. String Matching")
print("String matching is the process of finding a pattern inside a text.")
print('Example: Finding "ACG" inside "ACGTACG"')
```

```python
# QUESTION 3 — NAIVE PATTERN MATCHING

print("=" * 60)
print("QUESTION 3 — NAIVE PATTERN MATCHING")
print("=" * 60)

T = "ACGTACGTACGT"
P = "TAC"

print("Text   :", T)
print("Pattern:", P)

n = len(T)
m = len(P)

occurrences = []

print("\nAlignments:")

for i in range(n - m + 1):

    print("\nAlignment starting at position", i + 1)

    print("Text section :", T[i:i + m])
    print("Pattern      :", P)

    match = True

    for j in range(m):

        if T[i + j] == P[j]:

            print(
                "Character",
                j + 1,
                ":",
                T[i + j],
                "=",
                P[j],
                "MATCH"
            )

        else:

            print(
                "Character",
                j + 1,
                ":",
                T[i + j],
                "!=",
                P[j],
                "MISMATCH"
            )

            match = False

    if match:

        print("Result: COMPLETE MATCH")
        occurrences.append(i + 1)

    else:

        print("Result: MISMATCH")


print("\nPattern occurs at positions:")
print(occurrences)
```

```python
# QUESTION 4 — LPS ARRAY

print("=" * 60)
print("QUESTION 4 — LPS ARRAY")
print("=" * 60)

P = "ABABCABAB"

print("Pattern:", P)

# Create LPS array
lps = [0] * len(P)

length = 0
i = 1

while i < len(P):

    if P[i] == P[length]:

        length = length + 1
        lps[i] = length
        i = i + 1

    elif length != 0:

        length = lps[length - 1]

    else:

        lps[i] = 0
        i = i + 1


# Display LPS table

print("\nLPS Table")
print("-" * 30)
print("Index | Character | LPS")
print("-" * 30)

for i in range(len(P)):

    print(
        i,
        "    |",
        P[i],
        "        |",
        lps[i]
    )


print("\nLPS Array:")
print(lps)
```

```python
# QUESTION 5 — KMP MATCHING

print("=" * 60)
print("QUESTION 5 — KMP MATCHING")
print("=" * 60)

T = "ABABDABABCABAB"
P = "ABABCABAB"

print("Text   :", T)
print("Pattern:", P)


# ------------------------------------------------------------
# STEP 1 — CREATE THE LPS ARRAY
# ------------------------------------------------------------

lps = [0] * len(P)

length = 0
i = 1

while i < len(P):

    if P[i] == P[length]:

        length = length + 1
        lps[i] = length
        i = i + 1

    elif length != 0:

        length = lps[length - 1]

    else:

        lps[i] = 0
        i = i + 1


print("\n1. LPS ARRAY:")
print(lps)


# ------------------------------------------------------------
# STEP 2 — KMP MATCHING
# ------------------------------------------------------------

print("\n2. MAJOR MATCHING STEPS:")

i = 0
j = 0

occurrences = []

while i < len(T):

    if T[i] == P[j]:

        print(
            "MATCH:",
            "Text[", i, "] =", T[i],
            "Pattern[", j, "] =", P[j]
        )

        i = i + 1
        j = j + 1

        # Complete pattern found
        if j == len(P):

            position = i - j

            occurrences.append(position)

            print(
                ">>> COMPLETE MATCH FOUND "
                "at 0-based position:",
                position
            )

            # Use LPS to continue searching
            j = lps[j - 1]

    else:

        print(
            "MISMATCH:",
            "Text[", i, "] =", T[i],
            "Pattern[", j, "] =", P[j]
        )

        if j != 0:

            print(
                "Using LPS value:",
                lps[j - 1]
            )

            j = lps[j - 1]

        else:

            i = i + 1


# ------------------------------------------------------------
# STEP 3 — FINAL RESULT
# ------------------------------------------------------------

print("\n3. PATTERN LOCATION:")

print("0-based position(s):", occurrences)

print(
    "1-based position(s):",
    [x + 1 for x in occurrences]
)
```

```python
# QUESTION 6 — SUFFIX ARRAYS

print("=" * 60)
print("QUESTION 6 — SUFFIX ARRAYS")
print("=" * 60)

S = "BANANA$"


# ------------------------------------------------------------
# 6(a) LIST ALL SUFFIXES
# ------------------------------------------------------------

print("\n6(a) All suffixes:")

suffixes = []

for i in range(len(S)):

    suffix = S[i:]

    suffixes.append((i, suffix))

    print(
        "Position",
        i,
        ":",
        suffix
    )


# ------------------------------------------------------------
# 6(b) SORT THE SUFFIXES LEXICOGRAPHICALLY
# ------------------------------------------------------------

print("\n6(b) Sorted suffixes:")

sorted_suffixes = sorted(
    suffixes,
    key=lambda x: x[1]
)

for position, suffix in sorted_suffixes:

    print(
        "Position",
        position,
        ":",
        suffix
    )


# ------------------------------------------------------------
# 6(c) CONSTRUCT THE SUFFIX ARRAY
# ------------------------------------------------------------

print("\n6(c) Suffix Array:")

suffix_array = []

for position, suffix in sorted_suffixes:

    suffix_array.append(position)

print(suffix_array)


# ------------------------------------------------------------
# 6(d) SEARCH FOR ANA
# ------------------------------------------------------------

print("\n6(d) Search for ANA:")

pattern = "ANA"

ana_positions = []

for position in suffix_array:

    suffix = S[position:]

    if suffix.startswith(pattern):

        ana_positions.append(position)

        print(
            "Match found at position",
            position,
            ":",
            suffix
        )


# Sort positions from smallest to largest
ana_positions.sort()

print("\nANA positions using 0-based indexing:")
print(ana_positions)

print("\nANA positions using 1-based indexing:")
print([x + 1 for x in ana_positions])


# ------------------------------------------------------------
# FINAL ANSWER FOR QUESTION 6
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL ANSWER — QUESTION 6")
print("=" * 60)

print("Suffix Array:", suffix_array)

print(
    "ANA occurs at 0-based positions:",
    ana_positions
)

print(
    "ANA occurs at 1-based positions:",
    [x + 1 for x in ana_positions]
)
```
