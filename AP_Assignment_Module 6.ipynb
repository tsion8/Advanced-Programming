```python
from IPython.display import display, Markdown

display(Markdown("""


### <span style="color:blue">a) Brute Force</span>

**Definition:**  
Solve a problem by trying every possible solution (or combination) in a straightforward, exhaustive manner, without any cleverness to skip unlikely candidates. It relies on raw computing power rather than insight.

**Example:**  
Trying every possible combination on a numeric padlock (e.g., 0000–9999) until the correct one opens it.

---

### <span style="color:blue">b) Divide and Conquer</span>

**Definition:**  
Break a problem into smaller subproblems of the same type, solve each subproblem recursively, and then combine their solutions to solve the original problem.

**Example:**  
Sorting a large stack of exam papers by splitting it into smaller stacks, sorting each smaller stack, and then merging the sorted stacks back together (Merge Sort).

---

### <span style="color:blue">c) Greedy Algorithm</span>

**Definition:**  
Build a solution step by step, at each step choosing the option that looks best right now (locally optimal), without reconsidering earlier choices, hoping this leads to a globally optimal solution.

**Example:**  
GPS navigation using Dijkstra's algorithm chooses the shortest known route at each step.

---

### <span style="color:blue">d) Dynamic Programming</span>

**Definition:**  
Solve a complex problem by breaking it into overlapping subproblems, solving each subproblem only once, and storing (memoizing) the results to avoid redundant computation. The stored results are then combined to build the final solution efficiently.

**Example:**  
Planning the shortest or cheapest route on a road trip with multiple possible stopover cities, where the best route to each city is built from previously computed optimal routes instead of recalculating every possibility.

---

### <span style="color:blue">e) Backtracking</span>

**Definition:**  
Build a solution incrementally, and whenever a partial solution violates the constraints of the problem, abandon ("backtrack" from) that path and try a different option. The algorithm explores the solution space like a tree, pruning branches that cannot lead to a valid answer.

**Example:**  
Solving a Sudoku puzzle by filling in one number at a time. If a later cell has no valid number, the algorithm goes back and changes an earlier choice until a valid solution is found.
"""))
```

```python
import random

# Global counter for recursive calls
recursive_calls = 0


def merge_sort(arr):
    """Sort a list using the Merge Sort algorithm."""

    global recursive_calls
    recursive_calls += 1

    # Base case
    if len(arr) <= 1:
        return arr

    # Divide
    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Conquer
    return merge(left, right)


def merge(left, right):
    """Merge two sorted lists into one sorted list."""

    merged = []
    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1

        else:
            merged.append(right[j])
            j += 1

    # Add remaining elements
    merged.extend(left[i:])
    merged.extend(right[j:])

    return merged


# ---------------- Main Program ----------------

original_array = [random.randint(1, 100) for _ in range(15)]

print("=" * 50)
print("MERGE SORT USING DIVIDE AND CONQUER")
print("=" * 50)

print("\nOriginal Array:")
print(original_array)

# Reset counter
recursive_calls = 0

sorted_array = merge_sort(original_array)

print("\nSorted Array:")
print(sorted_array)

print("\nTotal Recursive Calls:", recursive_calls)
```

## Interpretation of Recursive Calls

The variable `recursive_calls` counts how many times the `merge_sort()` function calls itself.

Each recursive call divides the array into smaller subarrays until reaching lists containing one element.

For an array of size n, Merge Sort creates approximately:

2n - 1 recursive calls

because every division creates two smaller branches until reaching the base case.

This demonstrates the Divide and Conquer strategy used by Merge Sort.

```python
def brute_force_motif_search(dna, motif):
    """
    Searches for all occurrences of a motif in a DNA sequence
    using the brute-force approach.
    """

    positions = []
    comparisons = 0

    n = len(dna)
    m = len(motif)

    # Check every possible starting position
    for i in range(n - m + 1):

        match = True

        # Compare motif characters one by one
        for j in range(m):

            comparisons += 1

            if dna[i + j] != motif[j]:
                match = False
                break

        if match:
            positions.append(i + 1)   # Biological 1-based position

    return positions, comparisons


# ---------------- Main Program ----------------

dna = "ATCGTAGCTAGCTAGCTA"
motif = "TAG"

positions, comparisons = brute_force_motif_search(dna, motif)

print("="*50)
print("BRUTE-FORCE DNA MOTIF SEARCH")
print("="*50)

print("\nDNA Sequence:", dna)
print("Motif:", motif)

print("\nMotif found at position(s):", positions)

print("Number of comparisons made:", comparisons)
```

## Interpretation of Motif Positions

In this program, the line:

`positions.append(i + 1)`

is used to convert Python's zero-based indexing into biological one-based indexing.

Python normally starts counting positions from 0. For example, the first nucleotide in a sequence has index 0.

However, in bioinformatics, DNA sequence positions are usually reported starting from position 1.

For example:

DNA sequence:
ATCGTAGCTAGCTAGCTA

Python indexing:
- T in TAG starts at positions 4, 8, and 12

Biological indexing:
- T in TAG starts at positions 5, 9, and 13

Therefore, adding `+1` makes the output easier to interpret biologically.

## Why BLAST is Faster Than Brute Force

The brute-force approach compares the motif against every possible position in the DNA sequence. 
For long DNA sequences, this requires many unnecessary comparisons and becomes computationally expensive.

BLAST (Basic Local Alignment Search Tool) is faster because it uses a heuristic approach:

- It first searches for short matching words (called seeds) instead of comparing the entire sequence.
- It only extends promising matches.
- It avoids searching regions that are unlikely to contain meaningful similarity.
- It uses indexing techniques to quickly locate similar sequences.

Therefore, BLAST reduces the number of comparisons and can efficiently search very large biological databases.

## 1. Backtracking Algorithm for Motif Search

A backtracking algorithm builds a possible motif step by step.

The algorithm starts with an empty sequence and adds one nucleotide at a time 
(A, T, C, or G). After adding each nucleotide, it checks whether the partial motif 
appears in all DNA sequences.

If the partial motif cannot produce a valid common motif, the algorithm abandons 
that branch and returns to the previous step to try another nucleotide.

This process continues until a common motif is identified.

```python
import matplotlib.pyplot as plt
import networkx as nx

# Create a search tree graph
G = nx.DiGraph()

# Add nodes and edges
edges = [
    ("Start", "A"),
    ("Start", "T"),
    ("Start", "C"),
    ("Start", "G"),
    ("A", "AT"),
    ("T", "TT"),
    ("C", "CT"),
    ("G", "GT"),
    ("AT", "ATC"),
    ("GT", "GTC"),
    ("ATC", "ATCG"),
    ("GTC", "GTCG"),
    ("ATCG", "ATCGT"),
    ("CT", "CTC"),
    ("TT", "TTC")
]

G.add_edges_from(edges)

# Positions
positions = {
    "Start": (0, 5),
    "A": (-3, 4),
    "T": (-1, 4),
    "C": (1, 4),
    "G": (3, 4),
    "AT": (-3, 3),
    "TT": (-1, 3),
    "CT": (1, 3),
    "GT": (3, 3),
    "ATC": (-3, 2),
    "CTC": (1, 2),
    "GTC": (3, 2),
    "TTC": (-1, 2),
    "ATCG": (-3, 1),
    "GTCG": (3, 1),
    "ATCGT": (-3, 0)
}

# Node colors
pruned_nodes = {"CTC", "TTC"}
success_node = {"ATCGT"}

node_colors = []

for node in G.nodes():
    if node in success_node:
        node_colors.append("#a5d6a7")
    elif node in pruned_nodes:
        node_colors.append("#ef9a9a")
    elif node == "Start":
        node_colors.append("#e0e0e0")
    else:
        node_colors.append("#90caf9")


plt.figure(figsize=(10,7))

nx.draw(
    G,
    pos=positions,
    with_labels=True,
    node_size=2500,
    font_size=10,
    node_color=node_colors,
    arrows=True
)

plt.title("Backtracking Search Tree for DNA Motif Discovery")

plt.show()
```

### Interpretation of the Search Tree

The tree represents the process of exploring possible DNA motifs using backtracking.

The algorithm starts from an empty motif and adds nucleotides step by step.
Each branch represents a possible motif candidate.

Blue nodes represent active search paths, where the algorithm continues exploring.
Red nodes represent pruned branches, where the partial motif cannot produce a valid common motif.
The green node represents a successful motif candidate.

Pruning reduces the search space by stopping exploration of impossible solutions early.

# 3. Where Pruning Occurs
Pruning occurs when the algorithm detects that a partial motif cannot appear 
in all DNA sequences.

For example:

- If the partial sequence "CTC" appears in only some sequences,
  the algorithm stops exploring longer motifs beginning with "CTC".
- There is no need to test extensions such as "CTCA", "CTCG", or "CTCT".

By removing impossible branches early, backtracking reduces the search space 
and performs fewer comparisons than brute force.
