```python
# ======================================================================
# QUESTION 1: Comparing Genome Representation Concepts
# ======================================================================

class GenomeConcept:
    def __init__(self, name, definition, structure, purpose):
        self.name = name
        self.definition = definition
        self.structure = structure
        self.purpose = purpose

    def display(self):
        print(f"\n{self.name}")
        print("-" * len(self.name))
        print(f"Definition : {self.definition}")
        print(f"Structure  : {self.structure}")
        print(f"Purpose    : {self.purpose}")


concepts = [
    GenomeConcept(
        name="Linear Reference Genome",
        definition="A single, consensus haploid sequence (e.g., GRCh38) "
                    "representing 'the' genome of a species",
        structure="Simple string/array of nucleotides",
        purpose="Standard coordinate system for alignment, annotation, "
                "variant calling"
    ),
    GenomeConcept(
        name="Assembly Graph",
        definition="A graph built DURING genome assembly from overlapping "
                    "reads or k-mers (e.g., De Bruijn graph, overlap graph)",
        structure="Nodes = sequences/k-mers, edges = overlaps or adjacency",
        purpose="Reconstructs an unknown genome sequence from raw "
                "sequencing reads"
    ),
    GenomeConcept(
        name="Variation Graph",
        definition="A graph encoding a reference sequence PLUS known "
                    "variants (SNPs, indels, structural variants) as "
                    "alternate paths",
        structure="Nodes = sequence segments, edges = allowed transitions, "
                   "paths = individual haplotypes",
        purpose="Enables alignment against multiple alleles simultaneously, "
                "reducing reference bias"
    ),
    GenomeConcept(
        name="Genome Graph",
        definition="An umbrella term for ANY graph structure used to "
                    "represent genomic sequence and its variation",
        structure="Varies by context (includes assembly, variation, "
                   "pangenome graphs)",
        purpose="General category — not a single specific data structure"
    ),
    GenomeConcept(
        name="Pangenome Graph",
        definition="A variation graph built from MANY complete genomes of "
                    "a species/population",
        structure="Nodes = shared/variant segments, paths = each "
                   "individual genome",
        purpose="Represents species-level diversity (core + accessory + "
                "unique genome content)"
    ),
]

print("=" * 70)
print("QUESTION 1: GENOME REPRESENTATION CONCEPTS")
print("=" * 70)
for c in concepts:
    c.display()

print("\nKey relationship:")
print("  Assembly graphs BUILD sequences from reads.")
print("  Variation graphs / pangenome graphs REPRESENT known diversity.")
print("  Pangenome graph = variation graph scaled to population level.")
print("  'Genome graph' = umbrella term covering all of the above.")
```

```python
# ======================================================================
# QUESTION 2: BFS and DFS Applications in Genomic Graph Analysis
# ======================================================================

class TraversalApplication:
    def __init__(self, algorithm, behavior, applications):
        self.algorithm = algorithm
        self.behavior = behavior
        self.applications = applications

    def display(self):
        print(f"\n{self.algorithm}")
        print("-" * len(self.algorithm))
        print(f"Behavior: {self.behavior}\n")
        print("Genomic Applications:")

        for i, app in enumerate(self.applications, start=1):
            print(f"   {i}. {app}")


# ----------------------------------------------------------------------
# BFS INFORMATION
# ----------------------------------------------------------------------

bfs_info = TraversalApplication(
    algorithm="Breadth-First Search (BFS)",
    behavior="Explores a graph level-by-level from a starting node",
    applications=[
        "Finding the shortest path between nodes in a genomic graph",
        "Identifying variants within a specified distance from a "
        "genomic region",
        "Exploring the local neighborhood around a genomic region "
        "of interest",
        "Finding nearby k-mers in a De Bruijn graph during local "
        "assembly or sequence extension"
    ]
)


# ----------------------------------------------------------------------
# DFS INFORMATION
# ----------------------------------------------------------------------

dfs_info = TraversalApplication(
    algorithm="Depth-First Search (DFS)",
    behavior="Explores as far as possible along one branch before "
              "backtracking",
    applications=[
        "Enumerating possible paths through branches in a variation "
        "or pangenome graph",
        "Detecting cycles in assembly graphs caused by repetitive "
        "genomic regions",
        "Traversing connected paths to explore possible contigs "
        "or haplotypes",
        "Identifying branches and rejoining structures such as "
        "bubbles caused by genomic variation"
    ]
)


# ----------------------------------------------------------------------
# DISPLAY RESULTS
# ----------------------------------------------------------------------

print("=" * 70)
print("QUESTION 2: BFS AND DFS IN GRAPH-BASED GENOMIC ANALYSIS")
print("=" * 70)

bfs_info.display()
dfs_info.display()


# ----------------------------------------------------------------------
# COMPARISON
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("BFS VS DFS")
print("=" * 70)

print("BFS -> explores level-by-level and is useful for finding "
      "shortest paths and nearby nodes.")

print("DFS -> explores deeply along one branch and is useful for "
      "path exploration, cycle detection, and branch analysis.")



```

```python
# ======================================================================
# QUESTION 3: Linear vs. Graph-Based Reference Genomes
# ======================================================================

class ReferenceType:
    def __init__(self, name, advantages, limitations):
        self.name = name
        self.advantages = advantages
        self.limitations = limitations

    def display(self):
        print(f"\n{self.name}")
        print("-" * len(self.name))

        print("Advantages:")
        for a in self.advantages:
            print(f"   + {a}")

        print("Limitations:")
        for l in self.limitations:
            print(f"   - {l}")


# ----------------------------------------------------------------------
# LINEAR REFERENCE GENOME
# ----------------------------------------------------------------------

linear_ref = ReferenceType(
    name="Linear Reference Genome",
    advantages=[
        "Simple and computationally efficient",
        "Compatible with widely used tools such as BWA and SAMtools",
        "Provides an easy coordinate system such as chromosome:position"
    ],
    limitations=[
        "Can cause reference bias because reads containing "
        "non-reference alleles may align poorly",
        "Does not naturally represent large insertions, deletions, "
        "structural variation, or extensive population diversity",
        "Usually represents one reference genome rather than the "
        "full genetic diversity of a species"
    ]
)


# ----------------------------------------------------------------------
# GRAPH-BASED REFERENCE GENOME
# ----------------------------------------------------------------------

graph_ref = ReferenceType(
    name="Graph-Based Reference Genome",
    advantages=[
        "Represents multiple alleles or haplotypes simultaneously, "
        "which can reduce reference bias",
        "Naturally represents SNPs, insertions, deletions, and "
        "structural variants as alternative paths",
        "Better represents genomic diversity across different "
        "individuals and populations",
        "Can improve read mapping and variant analysis in highly "
        "variable genomic regions"
    ],
    limitations=[
        "More computationally expensive than traditional linear "
        "genome alignment",
        "Graph-based coordinate systems can be more complex than "
        "simple chromosome:position coordinates",
        "Requires specialized tools and algorithms for graph-based "
        "alignment and analysis"
    ]
)


# ----------------------------------------------------------------------
# DISPLAY RESULTS
# ----------------------------------------------------------------------

print("=" * 70)
print("QUESTION 3: LINEAR VS. GRAPH-BASED REFERENCE GENOMES")
print("=" * 70)

linear_ref.display()
graph_ref.display()


# ----------------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)

print("Linear reference genomes -> simple, fast, and easy to analyze, "
      "but they represent genomic diversity poorly.")

print("Graph-based reference genomes -> more complex and computationally "
      "demanding, but they provide a more complete representation "
      "of genomic variation and diversity.")

```

```python
import networkx as nx
import matplotlib.pyplot as plt


# ======================================================================
# QUESTION 4: CONSTRUCT A DE BRUIJN GRAPH
# ======================================================================

def get_kmers(read, k):
    """Generate all k-mers from a sequencing read."""
    return [read[i:i+k] for i in range(len(read) - k + 1)]


def build_de_bruijn_graph(reads, k):
    """Build a directed De Bruijn graph."""
    G = nx.MultiDiGraph()
    all_kmers = []

    for read in reads:
        kmers = get_kmers(read, k)
        all_kmers.extend(kmers)

    for kmer in all_kmers:
        prefix = kmer[:-1]
        suffix = kmer[1:]
        G.add_edge(prefix, suffix, label=kmer)

    return G, all_kmers


# ======================================================================
# GIVEN READS
# ======================================================================

reads = ["ATGC", "TGCA", "GCAT", "CATG"]
k = 3

G, all_kmers = build_de_bruijn_graph(reads, k)


# ======================================================================
# 1. LIST ALL 3-MERS
# ======================================================================

print("=" * 70)
print("QUESTION 4: DE BRUIJN GRAPH")
print("=" * 70)

print("\n1. ALL 3-MERS")

print("All 3-mers:", all_kmers)
print("Unique 3-mers:", sorted(set(all_kmers)))


# ======================================================================
# 2. IDENTIFY ALL 2-MER NODES
# ======================================================================

print("\n2. 2-MER NODES")

print("Nodes:", sorted(G.nodes()))


# ======================================================================
# 3. CONSTRUCT AND DRAW THE DIRECTED DE BRUIJN GRAPH
# ======================================================================

print("\n3. DIRECTED DE BRUIJN GRAPH")

for u, v, data in G.edges(data=True):
    print(f"{u} --({data['label']})--> {v}")


# Draw the graph
plt.figure(figsize=(10, 5))

pos = {
    "AT": (0, 0),
    "TG": (2, 0),
    "GC": (4, 0),
    "CA": (6, 0)
}

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2500,
    arrows=True,
    font_size=12
)

# Add k-mer labels to the edges
edge_labels = {}

for u, v, data in G.edges(data=True):
    edge_labels[(u, v)] = data["label"]

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=10
)

plt.title("Directed De Bruijn Graph (k = 3)")
plt.axis("off")
plt.show()


# ======================================================================
# 4. CALCULATE IN-DEGREE AND OUT-DEGREE
# ======================================================================

print("\n4. IN-DEGREE AND OUT-DEGREE")

for node in sorted(G.nodes()):
    print(
        f"{node}: "
        f"in-degree = {G.in_degree(node)}, "
        f"out-degree = {G.out_degree(node)}"
    )


# ======================================================================
# 5. DETERMINE EULERIAN PATH OR CYCLE
# ======================================================================

print("\n5. EULERIAN PATH / CYCLE")

if nx.is_eulerian(G):
    print("Every node has equal in-degree and out-degree.")
    print("An Eulerian CYCLE exists.")

elif nx.has_eulerian_path(G):
    print("An Eulerian PATH exists, but there is no Eulerian cycle.")

else:
    print("Neither an Eulerian path nor an Eulerian cycle exists.")


# ======================================================================
# 6. RECONSTRUCT THE SEQUENCE
# ======================================================================

print("\n6. RECONSTRUCTED SEQUENCE")

if nx.is_eulerian(G):

    cycle = list(nx.eulerian_circuit(G))

    print("Eulerian cycle:")

    for u, v in cycle:
        print(f"{u} -> {v}")

    # Reconstruct the sequence
    sequence = cycle[0][0]

    for _, v in cycle:
        sequence += v[-1]

    print("\nReconstructed sequence:", sequence)

    print("Circular genome representation: ATGC")
```

```python
from collections import deque
import networkx as nx


# ======================================================================
# QUESTION 6: GRAPH TRAVERSAL
# ======================================================================

# Given directed graph
graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D", "E"],
    "D": [],
    "E": ["F"],
    "F": []
}


# ======================================================================
# BFS - BREADTH-FIRST SEARCH
# ======================================================================

def bfs(graph, start):
    visited = [start]
    queue = deque([start])
    order = [start]

    while queue:
        node = queue.popleft()

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.append(neighbor)
                order.append(neighbor)
                queue.append(neighbor)

    return order


# ======================================================================
# DFS - DEPTH-FIRST SEARCH
# ======================================================================

def dfs(graph, start, visited=None, order=None):

    if visited is None:
        visited = set()

    if order is None:
        order = []

    visited.add(start)
    order.append(start)

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            dfs(graph, neighbor, visited, order)

    return order


# ======================================================================
# FIND ALL PATHS
# ======================================================================

def all_paths(graph, start, end, path=None):

    if path is None:
        path = []

    path = path + [start]

    if start == end:
        return [path]

    paths = []

    for neighbor in graph.get(start, []):
        if neighbor not in path:
            paths.extend(
                all_paths(graph, neighbor, end, path)
            )

    return paths


# ======================================================================
# 1. BFS FROM A
# ======================================================================

print("=" * 70)
print("QUESTION 6: GRAPH TRAVERSAL")
print("=" * 70)

bfs_result = bfs(graph, "A")

print("\n1. BFS from A:")
print("   " + " -> ".join(bfs_result))


# ======================================================================
# 2. DFS FROM A
# ======================================================================

dfs_result = dfs(graph, "A")

print("\n2. DFS from A:")
print("   " + " -> ".join(dfs_result))


# ======================================================================
# 3. ALL PATHS FROM A TO D
# ======================================================================

print("\n3. All paths from A to D:")

paths = all_paths(graph, "A", "D")

for path in paths:
    print("   " + " -> ".join(path))


# ======================================================================
# 4. IDENTIFY BRANCHING NODES
# ======================================================================

branching_nodes = [
    node for node, neighbors in graph.items()
    if len(neighbors) > 1
]

print("\n4. Branching nodes:")
print("   " + ", ".join(branching_nodes))


# ======================================================================
# 5. DETERMINE WHETHER THE GRAPH IS ACYCLIC
# ======================================================================

G = nx.DiGraph(graph)

is_acyclic = nx.is_directed_acyclic_graph(G)

print("\n5. Is the graph acyclic?")
print("   ", is_acyclic)


# ======================================================================
# SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)

print("BFS: A -> B -> C -> D -> E -> F")
print("DFS: A -> B -> D -> C -> E -> F")
print("Paths A -> D:")
print("   A -> B -> D")
print("   A -> C -> D")
print("Branching nodes: A, C")
print("Acyclic: True")


```
